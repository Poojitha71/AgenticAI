```python
## closures

def main_welcome(msg):
    def sub_welcome_method():
        print("welcome to adv python class")
        print(msg)
        print("please focus and learn properly")
    return sub_welcome_method
```

```python
main_welcome("hey")()
```

```python
def main_welcome(func):
    def sub_welcome_method():
        print("welcome to adv python class")
        func("hey all")
        print("please focus and learn properly")
    return sub_welcome_method
```

```python
main_welcome(print)()
```

```python
def main_welcome(func,lst):
    def sub_welcome_method():
        print("welcome to adv python class")
        print(func(lst))
        print("please focus and learn properly")
    return sub_welcome_method
```

```python
main_welcome(len,[1,2,3,4,5])()
```

```python
## decorators
def main_welcome(func):
    def sub_main():
        print("hey")
        func()
        print("welcome")
    return sub_main()
```

```python
def course_intro():
    print("this is a good day!!")
```

```python
course_intro()
```

```python
main_welcome(course_intro)
```

```python
@main_welcome
def course_intro():
    print("this is a good day!!")
```

```python
def my_decorator(func):
    def sub_fun():
        print("hello")
        func()
        print("bye")
        
    return sub_fun()
```

```python
@my_decorator
def say_hello():
    print("nice to meet you!!!")
```

```python
## decorators with parameters
def repeat(n):
    def my_deco(func):
        def wrapper(*args,**kwargs):
            for _ in range(n):
                func(*args,**kwargs)
        return wrapper
    return my_deco
        
```

```python
@repeat(3)
def say_hello():
    print("hello")
```

```python
say_hello()
```
